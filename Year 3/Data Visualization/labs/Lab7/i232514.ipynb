```python
import networkx as nx
from pyvis.network import Network
import random

friends = nx.Graph()

#generate random numbers and connect the nodes
for i in range(1, 101):
    friends.add_edge(i, (i % 100) + 1, title=f"Connects {i} and {(i % 100) + 1}",color = 'blue')
   
    rand_friend = random.randint(1, 100)
    while rand_friend == i:
        rand_friend = random.randint(1, 100)
    friends.add_edge(i, rand_friend, title=f"Connects {i} and {rand_friend}")


degree_map = {node: friends.degree[node] for node in friends.nodes()}

most_connected_node = max(degree_map, key=degree_map.get)
most_connections = degree_map[most_connected_node]

print(f"The number of connections for each node: {degree_map}")
print(f"\nThe node with the most connections is '{most_connected_node}' with {most_connections} connections.")


net = Network(notebook=True, cdn_resources='remote')

net.from_nx(friends)

for node in net.nodes:
    if node['id'] == most_connected_node:
        node['color'] = 'red' 

net.show("friends.html")
```

```python

hierarchy = nx.DiGraph()

hierarchy.add_node("Rector", label="Rector: Dr. Aftab")
hierarchy.add_node("Dean", label="Dean Dr. Sadia")
hierarchy.add_node("HoD_CS", label="HoD CS: Dr. Hassan Mujtaba")
hierarchy.add_node("HoD_EE", label="HoD EE: Dr. Nadeem Ali")
hierarchy.add_node("Prof_CS1", label="Prof CS: Ms. Sara")
hierarchy.add_node("Prof_CS2", label="Prof CS: Mr. Bilal")
hierarchy.add_node("Prof_EE1", label="Prof EE: Dr. Ahmed")
hierarchy.add_node("Lect_CS", label="Lect CS: Ms. Ayesha")
hierarchy.add_node("Lab_Instructor", label="Lab Instructor: Ms Maryam")

hierarchy.add_edges_from([
    ("Rector", "Dean"),
    ("Dean", "HoD_CS"),
    ("Dean", "HoD_EE"),
    ("HoD_CS", "Prof_CS1"),
    ("HoD_CS", "Prof_CS2"),
    ("HoD_EE", "Prof_EE1"),
    ("Prof_CS1", "Lect_CS"),
    ("Lect_CS", "Lab_Instructor"),
])


in_degrees = hierarchy.in_degree()
out_degrees = hierarchy.out_degree()

for node in hierarchy.nodes():
    node_in_degree = in_degrees[node]
    node_out_degree = out_degrees[node]
    print(f"Node '{node}': In-degree = {node_in_degree}, Out-degree = {node_out_degree}")
    
   
    hierarchy.nodes[node]['label'] = f"{hierarchy.nodes[node]['label']}\nIn: {node_in_degree}, Out: {node_out_degree}"
    hierarchy.nodes[node]['title'] = f"In-degree: {node_in_degree}, Out-degree: {node_out_degree}"



longest_path_length = nx.dag_longest_path_length(hierarchy)
dean_to_prof_paths = [
    nx.dag_longest_path_length(nx.subgraph(hierarchy, nx.dfs_tree(hierarchy, source="Dean")))
]
hierarchy_depth = max(dean_to_prof_paths) + 1  
print(f"\n--- Hierarchy Depth ---")
print(f"Longest chain from Dean to a professor: {hierarchy_depth} levels.")


net = Network(height='750px', width='100%', bgcolor="#F9FFA3", font_color="#222222", notebook=True, directed=True)

net.from_nx(hierarchy)

net.set_options("""
var options = {
  "physics": {
    "hierarchicalRepulsion": {
      "centralGravity": 0.5,
      "nodeDistance": 180
    },
    "minVelocity": 0.75,
    "solver": "hierarchicalRepulsion",
    "stabilization": {
      "iterations": 1000
    }
  },
  "layout": {
      "hierarchical": {
          "enabled": true,
          "direction": "UD",
          "sortMethod": "directed"
      }
  }
}
""")


net.show('hierarchy_University_graph.html')

```

```python
import networkx as nx
from pyvis.network import Network

socialNetwork = nx.Graph()

socialNetwork.add_node(1, label="Aleena")
socialNetwork.add_node(2, label="Muaz")
socialNetwork.add_node(3, label="Amna")
socialNetwork.add_node(4, label="Sara")
socialNetwork.add_node(5, label="Ahmed")
socialNetwork.add_node(6, label="Bilal")
socialNetwork.add_node(7, label="Fatima")
socialNetwork.add_node(8, label="Hassan")

socialNetwork.add_edge(1, 2, title="Connects Aleena and Muaz")
socialNetwork.add_edge(2, 3, title="Connects Muaz and Amna")
socialNetwork.add_edge(3, 7, title="Connects Amna and Sara")
socialNetwork.add_edge(4, 5, title="Connects Sara and Ahmed")
socialNetwork.add_edge(5, 6, title="Connects Ahmed and Bilal")
socialNetwork.add_edge(6, 7, title="Connects Bilal and Fatima")
socialNetwork.add_edge(1, 7, title="Connects Aleena and Fatima")
socialNetwork.add_edge(3, 6, title="Connects Amna and Bilal")
socialNetwork.add_edge(4, 8, title="Connects Sara and Hassan")
socialNetwork.add_edge(1, 8, title="Connects Aleena and Hassan")

# the target user is aleena
aleenasID = 1


friends_of_target = set(socialNetwork.neighbors(aleenasID))


potential_friends = [
    node for node in socialNetwork.nodes() 
    if node != aleenasID and node not in friends_of_target
]


mutual_friends_count = {}
for potential_friend in potential_friends:
    mutual_count = len(list(nx.common_neighbors(socialNetwork, aleenasID, potential_friend)))
    if mutual_count > 0:
        mutual_friends_count[potential_friend] = mutual_count

print("Potential new friends and their mutual friend count with Aleena:")
for user_id, count in mutual_friends_count.items():
    print(f"  - User '{socialNetwork.nodes[user_id]['label']}': {count} mutual friends")

# sort in decreasing order
recommended_friends = sorted(
    mutual_friends_count,
    key=mutual_friends_count.get,
    reverse=True
)


top_recommendations = recommended_friends[:2]

print("\n--- Top 2 Friend Recommendations for Aleena ---")
for rec_id in top_recommendations:
    rec_name = socialNetwork.nodes[rec_id]['label']
    print(f"  - {rec_name} (with {mutual_friends_count[rec_id]} mutual friends)")


net = Network(notebook=True, cdn_resources='remote')

net.from_nx(socialNetwork)

for node in net.nodes:
    node_id = node['id']
    
    if node_id == aleenasID:
        node['color'] = 'red'  
        node['title'] = "Target User: Aleena"
    elif node_id in top_recommendations:
        node['color'] = 'pink' 
        node['title'] = f"Recommended: {node['label']} ({mutual_friends_count[node_id]} mutual friends)"
    elif node_id in friends_of_target:
        node['color'] = 'blue'  
        node['title'] = f"Friend: {node['label']}"
    else:
        node['color'] = 'lightblue'  
        node['title'] = f"User: {node['label']}"


net.show("socialnetwork.html")
```
