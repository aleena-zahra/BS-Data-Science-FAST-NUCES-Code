```python
# change date to datetime format
df2['Date'] = pd.to_datetime(df2['Date'])
```

```python
# quantity sales profit should be numeric
df2['Quantity'] = pd.to_numeric(df2['Quantity'], errors='coerce')
df2['Sales Amount'] = pd.to_numeric(df2['Sales Amount'], errors='coerce')
df2['Profit'] = pd.to_numeric(df2['Profit'], errors='coerce')
```

```python
# total sum = sum(sales) column
df2['Total Sum'] = df2['Sales Amount'].sum()
# profit margin = profit / sales
df2['Profit Margin'] = df2['Profit'] / df2['Sales Amount']
```

```python
# save dfs as excel files
df1.to_excel('Ecommerce_Sales_Data.xlsx', index=False)
df2.to_excel('Retail_Sales_Data.xlsx', index=False)
```
