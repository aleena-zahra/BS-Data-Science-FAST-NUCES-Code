```python

def clean_country(x):
    if pd.isna(x):
        return None

    x = x.strip()

    # rename first
    if x in rename_map:
        x = rename_map[x]

    # remove regions/income groups
    for key in remove_if_contains:
        if key.lower() in x.lower():
            return None

    # drop territories
    if x in territories:
        return None

    # keep only valid ISO names
    if x not in valid_countries:
        return None

    return x

```

```python
# fill socio economic group with country wise mode
# ['Eswatini' 'Micronesia' 'Monaco' 'North Korea' 'Tanzania']
developing_countries = [
    'Eswatini', 'Micronesia', 'North Korea', 'Tanzania'
]
developed_countries = [
    'Monaco'
]

def fill_socio_economic(group):
    mode_value = group['Socio Economic Group'].mode()
    if not mode_value.empty:
        group['Socio Economic Group'] = group['Socio Economic Group'].fillna(mode_value[0])
    else:
        country_name = group.name
        if country_name in developing_countries:
            group['Socio Economic Group'] = group['Socio Economic Group'].fillna('Developing')
        elif country_name in developed_countries:
            group['Socio Economic Group'] = group['Socio Economic Group'].fillna('Developed')

    return group
```

### Feature Engineering / Transformation

```python
# make new columns
# Disaster Exposure (E): Frequency and severity of disasters in a region.
# Vulnerability (V): Human and economic losses relative to population or GDP.
# Adaptive Capacity (A): Socio-economic and institutional strength for recovery.
df['Disaster Exposure (E)'] = (
    (df['DisNo. [EMDAT]'] / (df.groupby('Country Name')['Time'].transform('max') - df.groupby('Country Name')['Time'].transform('min') + 1))
    * (df['Total Damage, Adjusted (current US$) [EMDAT]'] / (df['GDP per capita, PPP (current international $) [NY.GDP.PCAP.PP.CD]'] * df['Urban population (% of total population) [SP.URB.TOTL.IN.ZS]']))
)
df['Vulnerability (V)'] = (
    (df['Total Deaths [EMDAT]'] + df['No. Affected [EMDAT]'] + df['No. Homeless [EMDAT]'] + df['No. Injured [EMDAT]'])
    / (df['Urban population (% of total population) [SP.URB.TOTL.IN.ZS]'] + 1)
)
df['Adaptive Capacity (A)'] = (
    (df['GDP per capita, PPP (current international $) [NY.GDP.PCAP.PP.CD]'] * df['Urban population (% of total population) [SP.URB.TOTL.IN.ZS]'] / 100)
    / (1 + df['Gini index [SI.POV.GINI]'])
)
```

```python
# check unique values of disaster type
df['Disaster Type'].unique()
```

```python
df["Disaster Severity Weight"] = df["Disaster Type"].map({
    "None": 0,
    "Drought": 1,
    "Epidemic": 2,
    "Extreme temperature": 3,
    "Flood": 3,
    "Landslide": 2,
    "Mass movement (dry)": 2,
    "Mass movement (wet)": 2,
    "Infestation": 1,
    "Glacial lake outburst flood": 2,
    "Impact":1,
    "Storm": 3,
    "Volcanic activity": 2,
    "Wildfire": 2,
    "Earthquake": 4,
    "Tsunami": 4,
    "Volcanic eruption": 2,
}).fillna(2)  # default weight for unknown types
```

```python
# more columns 

# DII = (F + Apop)/GDPpc × S
# F = fatalities per million people per year
# Apop = affected population (% of total population)
# GDPpc = GDP per capita
# S = severity weight (based on disaster type)

df['Disaster Impact Index(DII)'] = (df['Total Deaths [EMDAT]'] + (df['No. Affected [EMDAT]'] )) /( (df['GDP per capita, PPP (current international $) [NY.GDP.PCAP.PP.CD]'] + 1)* df['Disaster Severity Weight'])
```

```python

# RRS =
# (GDPgrowthpost − GDPgrowthpre )/Trecovery + (HDI + Govindex)/2
# Where:
# • GDPgrowthpost = GDP growth rate after disaster years
# • GDPgrowthpre = GDP growth rate before disaster
# • Trecovery = years to reach pre-disaster output
# • HDI = Human Development Index
# • Govindex = governance or institutional stability score

# Proxy GDP growth using YoY % change in GDP per capita (PPP)
gdp_pc_col = 'GDP per capita, PPP (current international $) [NY.GDP.PCAP.PP.CD]'
df['GDPpc_growth_%'] = df.groupby('Country Name')[gdp_pc_col].pct_change() * 100

# Pre/post growth within country
df['GDPgrowthpre'] = df.groupby('Country Name')['GDPpc_growth_%'].shift(1)
df['GDPgrowthpost'] = df.groupby('Country Name')['GDPpc_growth_%'].shift(-1)

# Approximate Trecovery: years until GDPpc returns to or exceeds previous year's level
def _trecovery(g):
    g = g.sort_values('Time')
    gdp = g[gdp_pc_col].to_numpy()
    times = g['Time'].to_numpy()
    prev = np.concatenate(([np.nan], gdp[:-1]))
    out = np.full(len(g), np.nan, dtype=float)
    for i in range(len(g)):
        if np.isnan(prev[i]) or np.isnan(gdp[i]):
            continue
        future_idx = np.where(gdp[i:] >= prev[i])[0]
        if future_idx.size:
            out[i] = max(int(times[i + future_idx[0]] - times[i]), 1)
    return pd.Series(out, index=g.index)

df['Trecovery'] = df.groupby('Country Name', group_keys=False).apply(_trecovery)
trec = df['Trecovery'].clip(lower=1).fillna(1)

# RRS
gov = df['Political Stability and Absence of Violence/Terrorism: Estimate [PV.EST]']
df['Resilience Recovery Score (RRS)'] = ((df['GDPgrowthpost'] - df['GDPgrowthpre']) / trec) + ((df['HDI'] + gov) / 2)
```

```python
# Composite Resilience Index (CRI)
#CRI =A/(E × V)
df["Composite Resilience Index (CRI)"] = df['Adaptive Capacity (A)']/(df["Disaster Exposure (E)"]*df["Vulnerability (V)"])
```
